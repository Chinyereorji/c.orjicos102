```python
import tkinter as tk
from tkinter import messagebox
import csv
import random
from datetime import datetime as time

tasks = ["Loading","Transporting","Reviewing Orders","Customer Service","Delivering Items"]

class Employee:
    def __init__(self, name):
        self.name = name
    def check_employee(self):
        with open("employee.csv", "r") as file:
            worker = csv.reader(file)
            next(worker)
            for row in worker:
                if row[0].strip().lower() == self.name.strip().lower():
                    return True
        return False
    def take_attendance(self, task):
        check_time= time.now()
        date_time =check_time.strftime("%Y-%m-%d %H:%M:%S")
        with open("attendance.csv", "a", newline="") as file:
            attendance = csv.writer(file)
            attendance.writerow([self.name, "Present", task, date_time])
            
    def assign_task(self):
        return random.choice(tasks)
    def refuse_access(self):
         return "Access denied! You are not an employee."


def submit():
    name = entry.get()
    emp = Employee(name)

    if emp.check_employee():
        task = emp.assign_task()
        emp.take_attendance(task)
        messagebox.showinfo("Success", f"{name} marked present\nTask: {task}")
    else:
        messagebox.showerror("Error", emp.refuse_access())


# GUI
root = tk.Tk()
root.title("Employee System")
root.geometry("300x200")

tk.Label(root, text="Enter your name:").pack(pady=10)

entry = tk.Entry(root)
entry.pack(pady=5)

tk.Button(root, text="Submit", command=submit).pack(pady=20)

root.mainloop()
```

```python
import tkinter as tk
from tkinter import messagebox



class DeliveryService:
    def __init__(self, location, weight):
        self.location = location
        self.weight = weight

    def calculate_charge(self):
        if self.location == "PAU":
            if self.weight >= 10:
                return 2000
            else:
                return 1500

        elif self.location == "Epe":
            if self.weight >= 10:
                return 5000
            else:
                return 4000

        else:
            return None


# GUI Class
class DeliveryApp:
    def __init__(self, root):
        self.root = root
        self.root.title("Delivery Charge Calculator")
        self.root.geometry("400x300")
        self.root.config(bg="lightblue")

        # Heading
        title = tk.Label(root, text="Delivery Service Payment Checker",
                         font=("Arial", 16, "bold"),
                         bg="lightblue")
        title.pack(pady=10)

       
        tk.Label(root, text="Select Location:",
                 font=("Arial",12),
                 bg="lightblue").pack()

        self.location_var = tk.StringVar()
        self.location_var.set("PAU")

        locations = ["PAU", "Epe"]
        location_menu = tk.OptionMenu(root, self.location_var, *locations)
        location_menu.pack(pady=10)

       
        tk.Label(root, text="Enter Package Weight (kg):",
                 font=("Arial",12),
                 bg="lightblue").pack()

        self.weight_entry = tk.Entry(root)
        self.weight_entry.pack(pady=10)

        # Button
        calc_button = tk.Button(root,
                                text="Calculate Fee",
                                command=self.show_charge,
                                bg="green",
                                fg="white",
                                font=("Arial",12,"bold"))
        calc_button.pack(pady=15)

       
        self.result_label = tk.Label(root,
                                     text="",
                                     font=("Arial",14,"bold"),
                                     bg="lightblue")
        self.result_label.pack(pady=10)

    def show_charge(self):
        try:
            weight = float(self.weight_entry.get())
            location = self.location_var.get()

            delivery = DeliveryService(location, weight)
            fee = delivery.calculate_charge()

            self.result_label.config(
                text=f"Amount to Pay: ₦{fee}"
            )

        except ValueError:
            messagebox.showerror("Input Error",
                                 "Please enter a valid weight.")


root = tk.Tk()
app = DeliveryApp(root)
root.mainloop()
```
