```python
import tkinter as tk
from tkinter import messagebox

class Zenith:
    def unique_services(self):
        pass

    def mutual_services(self):
        return []


class RetailBanking(Zenith):
    def unique_services(self):
        return [
            "Retirement and Education Accounts",
            "Loans and Mortgages",
            "Checking and Saving Accounts"
        ]

    def mutual_services(self):
        return [
            "Lines of Credit",
            "Investment Management and Accounts",
            "Insurance"
        ]


class GlobalBanking(Zenith):
    def unique_services(self):
        return [
            "Multi-Currency Management Services and Products",
            "Foreign Currency Accounts",
            "Foreign Currency Credit Cards",
            "Transborder Advisory Services",
            "Liquidity Management"
        ]

    def mutual_services(self):
        return []


class CommercialBanking(Zenith):
    def unique_services(self):
        return [
            "Advisory Services"
        ]

    def mutual_services(self):
        return [
            "Lines of Credit",
            "Investment Management and Accounts",
            "Insurance"
        ]



def display_services():
    employee = employee_entry.get()
    division = division_entry.get().lower()

    if division == "retail":
        bank = RetailBanking()
    elif division == "global":
        bank = GlobalBanking()
    elif division == "commercial":
        bank = CommercialBanking()
    else:
        messagebox.showerror("Error", "Invalid Division")
        return

    services = bank.unique_services() + bank.mutual_services()

    result = f"Employee: {employee}\n"
    result += f"Division: {division.capitalize()} Banking\n\n"
    result += "Services Rendered:\n"

    for s in services:
        result += f"• {s}\n"

    output_label.config(text=result)



root = tk.Tk()
root.title("Zenith Bank Services")
root.geometry("600x500")


tk.Label(root, text="Employee Name").pack(pady=5)
employee_entry = tk.Entry(root, width=40)
employee_entry.pack()


tk.Label(root, text="Division (Retail, Global, Commercial)").pack(pady=5)
division_entry = tk.Entry(root, width=40)
division_entry.pack()


tk.Button(root, text="Display Services",
          command=display_services).pack(pady=10)


output_label = tk.Label(root, text="", justify="left")
output_label.pack(pady=10)

root.mainloop()
```
